```python
import matplotlib.pyplot as plt
hours = [2,3,4,5,6,7,8]
marks = [50,70,75,80,90,95,99]
plt.scatter(hours,marks,color="Red",marker=".",s=100,label="students")
plt.title("Hours studied vs Marks Obtained")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
hours = [2,3,4,5,6,7,8]
marks = [50,70,75,80,90,95,99]
plt.scatter(hours,marks,color="Blue",marker=".",s=100)
plt.title("Hours studied vs Marks Obtained")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")
plt.plot(hours,marks,color="Red",label="lineplot")
plt.grid(True)
plt.legend()
plt.show()
```
